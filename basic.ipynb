# HTTP Requests — GET and POST

````python
import requests

# GET — fetch data, params go in URL as ?key=value
response = requests.get(
    "https://api.example.com/users",
    headers={"Content-Type": "application/json"},
    # params={"page": 1, "limit": 10}
)
print(response.status_code)
# print(response.json())
````

````python
# POST with JSON body — use json= (auto sets Content-Type: application/json)
response = requests.post(
    "https://api.example.com/users",
    json={"name": "Alice", "age": 30}
)
print(response.status_code)
# print(response.json())
````

````python
# POST with form data
response = requests.post(
    "https://api.example.com/login",
    headers={"Content-Type": "application/x-www-form-urlencoded"},
    data={"username": "alice", "password": "secret"}
)
print(response.status_code)
````

**Common content types:**

| Content-Type | Use | requests param |
|---|---|---|
| `application/json` | JSON body | `json=` |
| `application/x-www-form-urlencoded` | Form fields | `data=` |
| `multipart/form-data` | File uploads | `files=` |

> When you use `json=`, requests sets `Content-Type: application/json` automatically. same for data= and file=


# Regex

**Basic symbols:**

| Symbol | Meaning | Example |
|---|---|---|
| `.` | Any character except newline | `a.c` → abc, a1c |
| `*` | 0 or more | `ab*` → a, ab, abb |
| `+` | 1 or more | `ab+` → ab, abb |
| `?` | 0 or 1 (optional) | `colou?r` → color, colour |
| `^` | Start of string | `^hello` |
| `$` | End of string | `world$` |
| `\d` | Digit `[0-9]` | `\d+` → 123 |
| `\w` | Word char `[a-zA-Z0-9_]` | `\w+` → hello_1 |
| `\s` | Whitespace | `\s+` |
| `[abc]` | Any of a, b, c | `[aeiou]` |
| `[^abc]` | Not a, b, c | `[^0-9]` |
| `{n,m}` | Between n and m times | `\d{2,4}` |
| `{n}` | n times | `\d{4}` |
| `(a\|b)` | a or b | `cat\|dog` |
| `()` | Group | `(ab)+` |

````python
import re

# Email validator
def is_valid_email(email):
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w{2,}$'
    return bool(re.match(pattern, email))

emails = ["user@example.com", "bad@", "name.last@domain.co.uk", "@no-user.com"]
for e in emails:
    print(f"{e:<30} → {is_valid_email(e)}")
````

````python
# URL validator
def is_valid_url(url):
    pattern = r'^https?://[\w\.-]+(:\d+)?(/[\w\./\-?=&#%]*)?$'
    return bool(re.match(pattern, url))

urls = [
    "https://www.google.com",
    "http://api.site.com/users?id=1&page=2",
    "ftp://wrong.com",
    "https://",
    "http://localhost:8080/path",
]
for u in urls:
    print(f"{u:<45} → {is_valid_url(u)}")
````

**Email pattern breakdown:** `^[\w\.-]+@[\w\.-]+\.\w{2,}$`
```
[\w\.-]+   → one or more word chars, dots, hyphens  (local part)
@          → literal @
[\w\.-]+   → domain name
\.         → literal dot
\w{2,}     → TLD at least 2 chars (com, uk, io...)
```

**URL pattern breakdown:** `^https?://[\w\.-]+(:\d+)?(/...)?$`
```
https?     → http or https
://        → literal
[\w\.-]+   → domain
(:\d+)?    → optional port  :8080
(/...)?    → optional path/query
```
